# Real blind DNS turbulence-closure experiment (JHTDB)

Discovery uses real JHTDB `isotropic1024coarse` DNS; sealed holdout uses the separate real `isotropic4096` DNS. Atlas receives no named turbulence-closure catalogue.


## 1. Download the frozen real DNS inputs
The downloader requests only raw gridded velocity from the official JHTDB REST cutout endpoint and writes six local `.npz` files plus SHA-256 provenance.


```python
import subprocess, sys
subprocess.run([sys.executable, '-m', 'evaluation.fetch_jhtdb_real_snapshots'], check=True)
print('real manifest: examples/turbulence_dns_closure_manifest.real.json')
```

## 2. Freeze and run Atlas
The manifest was constructed only after the raw bytes were SHA-256 bound. Discovery is `isotropic1024coarse` (`R_lambda≈433`); sealed holdout is `isotropic4096` (`R_lambda=610.57`).


```python
import subprocess, sys
manifest='examples/turbulence_dns_closure_manifest.real.json'
cmd=[sys.executable,'-m','evaluation.turbulence_dns_closure_experiment','--manifest',manifest,'--components','x,y,z','--output','reports/TURBULENCE_DNS_CLOSURE_CURRENT.json','--summary']
subprocess.run(cmd, check=True)
```

## 3. Interpret the receipt
`PASS_PROTOCOL_INTEGRITY` means the blind/provenance protocol was respected. It does **not** mean a closure was discovered. Scientific outcomes are deliberately separate: `TRANSFER_CANDIDATE_SURVIVES_CURRENT_SEALED_AND_NULL_EVIDENCE_NOT_LAW`, `TRANSFER_FIT_SURVIVES_BUT_NULL_CONTROL_NOT_REJECTED`, or `REPRESENTATION_GAP_OR_TRANSFER_FAILURE`.


```python
import json
from pathlib import Path
p=Path('reports/TURBULENCE_DNS_CLOSURE_CURRENT.json')
if p.exists():
    r=json.loads(p.read_text())
    print(r['protocol_status'], r['protocol_passed'], '/', r['protocol_total'])
    for row in r['component_outcomes']:
        print(row)
```
