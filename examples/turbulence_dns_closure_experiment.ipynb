# Blind DNS turbulence-closure experiment

Atlas receives only masked resolved velocity fields, the filter width and an attested coarse-graining residual. No named turbulence-closure catalogue or precomputed derivative predictors are supplied.


## 1. DNS input contract
Each `.npz` snapshot must contain three same-shape 3D arrays `u`, `v`, `w` and either scalar `dx` (isotropic grid), scalar `dx`,`dy`,`dz`, or 1D coordinate arrays `x`,`y`,`z`. The current adapter assumes periodic DNS for the spectral coarse-graining target.


```python
from pathlib import Path
from evaluation.turbulence_dns_closure_experiment import _write_template
_write_template(Path('examples/turbulence_dns_closure_manifest.local.json'))
print('Edit examples/turbulence_dns_closure_manifest.local.json and point it to your DNS snapshots.')
```

## 2. Freeze and run
Use different `regime_id` values for discovery and sealed holdout. A different Reynolds number, forcing regime, or flow family is appropriate. The harness refuses overlapping discovery/sealed regime IDs.


```python
import subprocess, sys
manifest='examples/turbulence_dns_closure_manifest.local.json'
cmd=[sys.executable,'-m','evaluation.turbulence_dns_closure_experiment','--manifest',manifest,'--components','x,y,z','--output','reports/TURBULENCE_DNS_CLOSURE_CURRENT.json','--summary']
print(' '.join(cmd))
# subprocess.run(cmd, check=True)  # uncomment after paths in the manifest are valid
```

## 3. Interpret the receipt
`PASS_PROTOCOL_INTEGRITY` means the blind/provenance protocol was respected. It does **not** mean a closure was discovered. Scientific outcomes are deliberately separate: `TRANSFER_CANDIDATE_SURVIVES_CURRENT_SEALED_AND_NULL_EVIDENCE_NOT_LAW`, `TRANSFER_FIT_SURVIVES_BUT_NULL_CONTROL_NOT_REJECTED`, or `REPRESENTATION_GAP_OR_TRANSFER_FAILURE`.


```python
import json
from pathlib import Path
p=Path('reports/TURBULENCE_DNS_CLOSURE_CURRENT.json')
if p.exists():
    r=json.loads(p.read_text())
    print(r['protocol_status'], r['protocol_passed'], '/', r['protocol_total'])
    for row in r['component_outcomes']:
        print(row)
```
